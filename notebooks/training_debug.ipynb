```python
import sys
sys.path.insert(0, '../backend')

import torch
from model import create_model
from config import DEVICE, NUM_CLASSES

# Test model
model = create_model(arch='resnet18', num_classes=NUM_CLASSES, pretrained=False)
x = torch.randn(1, 3, 224, 224)
out = model(x)
print(f'Input: {x.shape}')
print(f'Output: {out.shape}')
print(f'Predictions: {torch.softmax(out, dim=1)}')
```

```python
# Load and visualize training history
import json
import matplotlib.pyplot as plt

with open('../backend/models/training_history.json') as f:
    history = json.load(f)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.plot(history['train_loss'])
ax1.set_title('Training Loss')
ax2.plot(history['train_acc'], label='Train')
ax2.plot(history['val_acc'], label='Val')
ax2.set_title('Accuracy')
ax2.legend()
plt.tight_layout()
plt.show()
```
