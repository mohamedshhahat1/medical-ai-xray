# 🔬 Data Exploration — Chest X-Ray Dataset

This notebook explores the X-ray dataset:
- Sample visualization
- Class distribution
- Image statistics (size, intensity)
- Quality assessment

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from pathlib import Path

DATA_DIR = '../data/raw'

# Count images per class
for split in ['train', 'val']:
    split_dir = os.path.join(DATA_DIR, split)
    if os.path.exists(split_dir):
        for cls in os.listdir(split_dir):
            cls_dir = os.path.join(split_dir, cls)
            if os.path.isdir(cls_dir):
                count = len(os.listdir(cls_dir))
                print(f'{split}/{cls}: {count} images')
```

```python
# Visualize sample images
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
fig.suptitle('Sample X-Ray Images', fontsize=14)
# TODO: Load and display samples
plt.tight_layout()
plt.show()
```
